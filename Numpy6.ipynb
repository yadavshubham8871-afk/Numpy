```python
# orders = np.array([
# [1000, 2, 500],
# [1001, 3, 750],
# [1002, 1, 1200],
# [1003, 4, 300],
# [1004, 2, 900]
# ])
# Columns:
# Column 1 → Order ID
# Column 2 → Quantity
# Column 3 → Price per item
# Tasks
# Display the complete information of Order 1002.
# Display all quantities.
# Display all prices.
# Display the first 3 orders.
# Display the last 2 orders.
# Increase the price of the last 2 products by 10%.
# Increase the quantity of the first order by 1.
# Display the final dataset.
# Create a separate array containing only the quantity and price columns.
# Display the shape and size of the data

orders = np.array([
[1000, 2, 500],
[1001, 3, 750],
[1002, 1, 1200],
[1003, 4, 300],
[1004, 2, 900]
])
print(orders[2])
print(orders[:,1])
print(orders[:,2])
print(orders[:3])
print(orders[-2:])
orders[-2:,2]=orders[-2:,2]*1.10
orders[0:1,1]=orders[0:1,1]+1
print(orders)
New_orders=orders[:,-2:]
print(New_orders)
print(orders.shape)
print(orders.size)
```

```python
# A simple grayscale image can be represented as a 2D NumPy array where each number represents pixel intensity.
# image = np.array([
# [10, 20, 30, 40, 50],
# [20, 30, 40, 50, 60],
# [30, 40, 50, 60, 70],
# [40, 50, 60, 70, 80],
# [50, 60, 70, 80, 90]
# ])
# Tasks
# Find the dimensions of the image.
# Find its shape.
# Find the total number of pixels.
# Display the center pixel.
# Extract the top-left 3 × 3 region.
# Extract the bottom-right 2 × 2 region.
# Increase the intensity of the center 3 × 3 region by 20.
# Display the modified image.

image = np.array([
[10, 20, 30, 40, 50],
[20, 30, 40, 50, 60],
[30, 40, 50, 60, 70],
[40, 50, 60, 70, 80],
[50, 60, 70, 80, 90]
])
print(image.ndim)
print(image.shape)
print(image.size)
print(image[2:3,2:3])
print(image[0:3,0:3])
print(image[3:,3:])
image[1:4,1:4]+=20
print(image)
```

```python
# A company stores quarterly revenue and expenses:
# financials = np.array([
# [120, 80],
# [150, 90],
# [180, 110],
# [200, 130]
# ])
# Rows = Q1, Q2, Q3, Q4
# Columns = Revenue, Expense
# Tasks
# Display Q2's complete financial data.
# Display all quarterly revenues.
# Display all quarterly expenses.
# Extract Q2 and Q3 data.
# Increase Q4 revenue by 15%.
# Reduce Q1 expense by 10%.
# Display the final financial matrix.

financials = np.array([
[120, 80],
[150, 90],
[180, 110],
[200, 130]
])
print(financials[1])
print(financials[:,0])
print(financials[:,1])
print(financials[1:3])
financials[3:,0]=financials[3:,0]*1.15
financials[0:1,1:2]=financials[0:1,1:2]*0.90
print(financials)

```

```python
# A classroom has 6 rows and 5 seats per row.
# Initially, all seats are empty.
# 0 = Empty
# 1 = Occupied
# Tasks
# Create a 6 × 5 array filled with zeros.
# Mark the first seat of the first row as occupied.
# Mark all seats in the 3rd row as occupied.
# Mark the last two seats of the last row as occupied.
# Display the first 3 rows.
# Display the last 2 columns.
# Display the complete seating matrix.

students=np.full((6,5),0)
print(students)
students[0:1,0:1]+=1
students[2:3]+=1
students[-1:,-2:]+=1
print(students[0:3])
print(students[:,-2:])
print(students)

```
