```python
# A college records marks of 5 students in 4 subjects.
# marks = np.array([
# [72, 85, 90, 78],
# [65, 70, 75, 80],
# [88, 92, 95, 90],
# [55, 65, 60, 70],
# [78, 82, 85, 88]
# ])
# Each row represents a student and each column represents a subject.
# Tasks
# Determine the number of dimensions.
# Determine the number of students and subjects using shape.
# Determine the total number of marks using size.
# Display the complete marks of the 3rd student.
# Display marks of all students in the 2nd subject.
# Extract marks of the first 3 students in the first 2 subjects.
# Increase the marks of the last 2 students in the 4th subject by 5

marks = np.array([
[72, 85, 90, 78],
[65, 70, 75, 80],
[88, 92, 95, 90],
[55, 65, 60, 70],
[78, 82, 85, 88]
])
print(marks.ndim)
print(marks.shape)
print(marks.size)
print(marks[2])
print(marks[:,1])
print(marks[:3,:2])
marks[3:,3]=marks[3:,3]+5
print(marks)

```

```python
# A restaurant stores prices of 6 menu items:
# prices = np.array([120, 180, 250, 300, 450, 600])
# The restaurant wants to run different pricing experiments.
# Tasks
# Add ₹20 to the first two items.
# Give a 10% discount to items 3, 4 and 5.
# Keep the last item unchanged.
# Display the first 3 prices after modification.
# Display the last 2 prices.
# Create a separate array containing only the middle three prices

prices = np.array([120, 180, 250, 300, 450, 600])
prices[:3]=prices[:3]+20
prices[2:6]=prices[2:6]*0.90
print(prices[:3])
print(prices[-2:])
New_prices=prices[2:5]
print(New_prices)

```

```python
# A machine records temperature every hour:
# temperature = np.array([
# 28, 29, 30, 31, 33, 35,
# 36, 35, 34, 32, 30, 29
# ])
# Tasks
# Extract readings from the first 6 hours.
# Extract readings from the last 6 hours.
# Extract readings from every alternate hour.
# Extract readings from every third hour.
# Reverse the complete temperature sequence.
# Extract the last 4 readings in reverse order.

temperature = np.array([
28, 29, 30, 31, 33, 35,
36, 35, 34, 32, 30, 29
])
print(temperature[:6])
print(temperature[-6:])
print(temperature[::2])
print(temperature[::3])
Reverse_temp=temperature[::-1]
print(Reverse_temp[-4::])

```

```python
# A warehouse has 4 storage sections and 5 products.
# Initially, every product has a stock level of 100 units.
# Tasks
# Create a 4 × 5 array where every value is 100.
# Increase the stock of the first two sections by 20 units.
# Reduce stock of the last two products by 10%.
# Display only the middle two sections.
# Display only the first three products.
# Display the bottom-right 2 × 2 portion of the matrix

units=np.full((4,5),100)
print(units)
units[:2]=units[:2]+20
units[:,-2:]=units[:,-2:]*0.90
print(units)
print(units[1:3])
print(units[:,:3])
print(units[-2:,-2:])
```

```python
# A company has salary data for 4 employees:
# salary = np.array([
# [30000, 5000],
# [40000, 6000],
# [50000, 8000],
# [60000, 10000]
# ])
# Column 1 = Basic Salary
# Column 2 = Allowance
# Tasks
# Display the complete information of the 2nd employee.
# Display all basic salaries.
# Display all allowances.
# Increase all allowances by ₹1,000.
# Increase the basic salary of the last two employees by 10%.
# Display the final salary matrix

salary = np.array([
[30000, 5000],
[40000, 6000],
[50000, 8000],
[60000, 10000]
])
print(salary[1:2])
print(salary[:,0])
print(salary[:,1])
salary[:,1]=salary[:,1]+1000
salary[-2:,:0]=salary[-2:,:0]*1.10
print(salary)
```
