```python
# A teacher has recorded marks of 10 students:
# marks = np.array([72, 85, 91, 64, 78, 88, 95, 70, 82, 76])
# Find:
# Total marks
# Average marks
# Highest marks
# Lowest marks
# Difference between highest and lowest marks
# Index of the student with the highest mark
marks = np.array([72, 85, 91, 64, 78, 88, 95, 70, 82, 76])
Total=0
max=marks[0]
min=marks[0]
#max_index=0
for mark in marks:
  Total+=mark
Average=Total/len(marks)
for i in range(len(marks)):
  if marks[i]>max:
    max=marks[i]
    max_index=i
  if marks[i]<min:
    min=marks[i]
    min_index=i
print("Total marks=",Total)
print("Average marks=",Average)
print("Highest marks=",max)
print(max_index)
print("Lowest marks=",min)
print(min_index)
print("Difference=",max-min)
```

```python
print(np.max(marks))
print(np.min(marks))
print(np.sum(marks))
print(np.mean(marks))
print(np.argmax(marks))
print(np.argmin(marks))
print(np.max(marks)-np.min(marks))
```

```python
# A company's daily sales for 7 days are:
# sales = np.array([45000, 47000, 49000, 52000, 50000, 48000, 250000])
# Calculate the mean and median.
# Then determine which measure better represents a typical sales day when considering the unusually large sale.
# Also calculate the standard deviation.

sales = np.array([45000, 47000, 49000, 52000, 50000, 48000, 250000])
print("Average sales=",round(np.mean(sales)))
print("Median sales=",np.median(sales))
print("Standard Deviation=",np.std(sales))

```

```python
# A business records revenue for 12 months:
# revenue = np.array([
# 120000, 135000, 128000, 145000,
# 150000, 138000, 160000, 172000,
# 155000, 148000, 180000, 195000
# ])
# Find:
# Average monthly revenue
# Highest revenue
# Lowest revenue
# Month index with highest revenue
# Month index with lowest revenue
# Total annual revenue
# Difference between highest and lowest revenue.

revenue = np.array([
120000, 135000, 128000, 145000,
150000, 138000, 160000, 172000,
155000, 148000, 180000, 195000
])
print("Average revenue=",np.mean(revenue))
print("Highest Revenue=",np.max(revenue))
print("Lowest Revenue=",np.min(revenue))
print("Month index with highest revenue:",np.argmax(revenue))
print("Month index with lowest revenue:",np.argmin(revenue))
print("Total annual revenue:",np.sum(revenue))
print("Difference B/W max and min:",np.max(revenue)-np.min(revenue))
```
