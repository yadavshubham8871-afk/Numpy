```python
# A company records sales for four quarters:
# Q1 = 125000
# Q2 = 148000
# Q3 = 135000
# Q4 = 172000
# Create a NumPy array.
# The company expects next year's sales to be 15% higher than this year's sales.
#Calculate the expected sales for each quarter.
# Then calculate the additional sales expected in each quarter

q=np.array([125000,148000,135000,172000])
i=q*0.15
ex_sales=i+q
add_sales=ex_sales-q
print(ex_sales)
print(add_sales)
```

```python
# Two branches of a company have inventory for five products.
# Branch A = [120, 150, 180, 100, 200]
# Branch B = [80, 130, 170, 120, 150]
# Create two NumPy arrays.
# Calculate:
# Total inventory for each product.
# Difference between Branch A and Branch B.
# Combined inventory after receiving an additional 20 units of every product.

a=np.array([120, 150, 180, 100, 200])
b=np.array([80, 130, 170, 120, 150])
d=a-b
print(a+b)
print(d)
print((a+20)+(b+20))
```

```python
#A company wants to set targets from ₹20,000 to ₹1,00,000.
#Generate exactly 9 target values, equally spaced.
#Then create actual sales:
#[18000, 28000, 39000, 47000, 57000, 66000, 76000, 90000, 105000]
#Calculate the difference between actual and target

target=np.linspace(20000,10000,9)
actual_sales=np.array([18000,28000,39000,47000,57000,66000,76000,90000,105000])
difference=actual_sales-target
print(target)
print(actual_sales)
print(difference)
```

```python
#A website offers:
#First discount: 10%
#Second discount: 5% on the already discounted price
# Prices:
# [1000, 2000, 3000, 4000]
# Calculate the final prices

price=np.array([1000, 2000, 3000, 4000])
print(price)
Fst_discount=price*0.1
New_price=price-Fst_discount
final_price=(New_price*0.95)
print(final_price)

```

```python
# A business has monthly revenue:
# [80000, 95000, 110000, 125000, 140000, 155000]
# The expected growth rate is 12%.
# Calculate:
# Expected increase.
# Forecast revenue.
# Create a target revenue array using linspace() from ₹80,000 to ₹1,60,000 with exactly 6 values.
# Compare forecast revenue with the target

revenue=np.array([80000, 95000, 110000, 125000, 140000, 155000])
print(revenue)
Exp_inc=revenue*0.12
Exp_revenue=Exp_inc+revenue
print(Exp_inc)
print(Exp_revenue)
target=np.linspace(80000,160000,6)
print(target)
compare=target-Exp_revenue
print(compare)
```
