```python
# The following array contains sales made by 4 salespeople across 3 months.
# sales = np.array([
# [120, 150, 180],
# [200, 170, 160],
# [100, 140, 130],
# [180, 190, 210]
# ])
# Rows represent salespeople and columns represent months
# Find:
# Total sales by each salesperson
# Average sales by each salesperson
# Total sales for each month
# Average sales for each month
# Highest sales achieved by each salesperson

sales = np.array([
[120, 150, 180],
[200, 170, 160],
[100, 140, 130],
[180, 190, 210]
])
print("Total sales by each salesperson:",np.sum(sales,axis=1))
print("Average sales by each salesperson:",np.round(np.mean(sales,axis=1),2))
print("Total sales for each month:",np.sum(sales,axis=0))
print("Total sales for each month",np.mean(sales,axis=0))
print("Highest sales achieved by each salesperson",np.max(sales,axis=1))
```

```python
# An electricity board records the monthly electricity consumption of a household for one year:
# consumption = np.array([
# 180, 220, 195, 250,
# 280, 310, 295, 270,
# 240, 210, 190, 230])
# Perform the following:
# Find the total annual consumption.
# Calculate average monthly consumption.
# Find the highest and lowest consumption.
# Find the month index with the highest consumption.
# Extract the first 6 months using slicing.
# Extract alternate months.
# Find all months where consumption was greater than the annual average.
# Calculate the new monthly projections if next year's consumption is expected to be 8% higher

consumption = np.array([
180, 220, 195, 250,
280, 310, 295, 270,
240, 210, 190, 230])
print("Total Consumption:",np.sum(consumption))
print("Average:",round(np.mean(consumption)))
print("Highest consumtion:",np.max(consumption))
print("Max_Index:",np.argmax(consumption))
print("First 6 months consumption:",consumption[:6])
print("Alternate Month consumption:",consumption[::2])
print("consumption was greater than the annual average",consumption[consumption>round(np.mean(consumption))])
print("Expected consumption:",consumption*1.08)


```

```python
# The following array contains marks of 5 students in 4 subjects:
# marks = np.array([
# [72, 85, 90, 78],
# [65, 70, 75, 80],
# [88, 92, 95, 90],
# [55, 60, 58, 65],
# [78, 82, 85, 88]
# ])
# Assume each row represents a student and each column represents a subject.
# Find:
# Shape and size of the array.
# Marks obtained by the third student.
# Marks obtained in the second subject by all students.
# Average marks of every student.
# Average marks of every subject.
# Students whose average is above 75.
# Highest mark obtained in each subject.
# The complete row of the student with the highest average.

marks = np.array([
[72, 85, 90, 78],
[65, 70, 75, 80],
[88, 92, 95, 90],
[55, 60, 58, 65],
[78, 82, 85, 88]
])
print("Shape and size of the array:",np.shape(marks),np.size(marks))
print("Marks of 3rd student:",marks[2:3])
print("Marks in second Subject:",marks[:,1])
print("Average marks of all student:",np.mean((marks),axis=1))
print("Average marks of all subject:",np.mean(marks,axis=0))
print("Students whose average is above 75:",marks[np.mean((marks),axis=1)>75])
print("Highest mark obtained in each subject:",np.max(marks,axis=0))
ind=np.argmax(np.mean((marks),axis=1))
print("The complete row of the student with the highest average:",marks[ind])


```

```python
# An online store has the following product prices:
# prices = np.array([
# 499.50, 799.90, 1299.75,
# 2499.50, 3999.90, 599.25,
# 899.80, 1599.40
# ])
# Perform:
# Round prices to the nearest whole number.
# Find products costing more than ₹1,000.
# Calculate a 15% discount only on products costing more than ₹1,000.
# Calculate the average discounted price of those products.
# Find the most expensive original product.
# Extract the last 4 products.
# Calculate the square root of the final prices and round them to 2 decimal places

prices = np.array([
499.50, 799.90, 1299.75,
2499.50, 3999.90, 599.25,
899.80, 1599.40
])
prices=np.round(prices)
print(prices)
print("Products costing more than ₹1,000:",prices[prices>1000])
Dis=prices[prices>1000]-(prices[prices>1000])*0.85
print(Dis)
print("Average discounted price:",np.mean(prices[prices>1000])*0.85)
print("Most expensive original product:",np.max(prices))
print("Index of max price:",np.argmax(prices))
print("Last 4 products:",prices[-4:])
print("square root of the final prices:",np.round(np.sqrt(prices),2))
```
